Cell K0.1, prepare the upload folder

```python
!pip install -q -U kaggle
import os, json, glob, shutil
from google.colab import drive, userdata
drive.mount("/content/drive")

os.environ["KAGGLE_API_TOKEN"] = userdata.get("KAGGLE_API_TOKEN")   # read from Secrets, never printed
KAGGLE_USER = "rahuldevagiri"                                # from kaggle.com/<username>
!kaggle datasets list --mine

OUT = "/content/drive/MyDrive/lulc_s1_global"
UP = "/content/kaggle_ds"; os.makedirs(UP, exist_ok=True)

files = [f"{OUT}/patches.zip"] + sorted(glob.glob(f"{OUT}/v2_parts/v2_part_[0-9]*.zip")) + [
    f"{OUT}/index_v2.csv", f"{OUT}/norm_stats_v2.json", f"{OUT}/norm_stats.json",
    f"{OUT}/exclude_pids.json", f"{OUT}/points_meta.csv", f"{OUT}/points_v2.json", f"{OUT}/v2_dropped.csv"]
for f in files:
    dst = f"{UP}/{os.path.basename(f)}"
    if not os.path.exists(dst):
        shutil.copy(f, dst)
    print(f"{os.path.basename(f):28s} {os.path.getsize(dst) / 1e9:6.2f} GB")

meta = {"title": "LULC Sentinel-1 Global v2",
        "id": f"{KAGGLE_USER}/lulc-s1-global-v2",
        "licenses": [{"name": "CC-BY-4.0"}]}
json.dump(meta, open(f"{UP}/dataset-metadata.json", "w"), indent=2)
print("files:", len(os.listdir(UP)))
```
